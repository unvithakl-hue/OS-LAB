```python
n = int(input("Enter number of processes: "))

bt = []

for i in range(n):
    time = int(input(f"Enter burst time for P{i+1}: "))
    bt.append(time)

q = int(input("Enter time quantum: "))

remaining = bt.copy()
wt = [0] * n
tat = [0] * n
time = 0

while True:
    done = True

    for i in range(n):
        if remaining[i] > 0:
            done = False

            if remaining[i] > q:
                time += q
                remaining[i] -= q
            else:
                time += remaining[i]
                wt[i] = time - bt[i]
                remaining[i] = 0

    if done:
        break

for i in range(n):
    tat[i] = bt[i] + wt[i]

print("\nProcess\tBurst Time\tWaiting Time\tTurnaround Time")

for i in range(n):
    print(f"P{i+1}\t{bt[i]}\t\t{wt[i]}\t\t{tat[i]}")

print("\nAverage Waiting Time =", sum(wt) / n)
print("Average Turnaround Time =", sum(tat) / n)
```
